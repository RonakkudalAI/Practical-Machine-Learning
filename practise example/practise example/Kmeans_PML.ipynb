```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
```

```python
X,y=load_iris(return_X_y=True)
```

```python
X
```

```python
y
```

```python
#cross validation method
kf=KFold(n_splits=5,shuffle=True,random_state=42)
```

```python
model=LogisticRegression()
```

```python
scores=[]
```

```python
for train_index,val_index in kf.split(X):
    X_train,X_val=X[train_index],X[val_index]
    y_train,y_val=y[train_index],y[val_index]

    model.fit(X_train,y_train)
    y_pred=model.predict(X_val)
    acc=accuracy_score(y_val,y_pred)
    scores.append(acc)

print("Accuracy",scores)
print("Average accuracy",np.mean(scores))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_blobs
from sklearn.metrics import accuracy_score
from sklearn.cluster import KMeans
```

```python
X,y=make_blobs(n_samples=300,centers=4,cluster_std=0.60,random_state=0)
```

```python
df=pd.DataFrame(X,columns=['feature1','feature2'])
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.scatter(df['feature1'],df['feature2'])
plt.title('Scatter plot')
plt.xlabel('feature1')
plt.ylabel('feature2')
plt.show()
```

```python
sc=StandardScaler()
X_sc=sc.fit_transform(df)
```

```python
y
```

```python
#eldow method
inertia=[]

for k in range(1,10):
  kmeans=KMeans(n_clusters=k,random_state=0)
  kmeans.fit(X_sc)
  inertia.append(kmeans.inertia_)
```

```python
plt.plot(range(1,10),inertia,marker='o')
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
```

```python
#apply model
kmeans=KMeans(n_clusters=4,random_state=0)
kmeans.fit(X_sc)
```

```python
df['Cluster']=kmeans.predict(X_sc)
```

```python
centers=kmeans.cluster_centers_
```

```python
plt.scatter(df['feature1'],df['feature2'],c=df['Cluster'],cmap='viridis',s=50)
center_originals=sc.inverse_transform(centers)
plt.scatter(center_originals[:,0],center_originals[:,1],c='red',s=200,marker='X')
plt.title('K-means clustering')
plt.xlabel('feature1')
plt.ylabel('feature2')
plt.show()
```

```python
from sklearn.metrics import silhouette_score
score=silhouette_score(X_sc,df['Cluster'])
print(score)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

```python
iris_data=load_iris(as_frame=True)#Loading full data
```

```python
iris=iris_data.frame
```

```python
iris['Species']=iris_data.target_names[iris_data.target]
```

```python
iris.shape
```

```python
iris.info()
```

```python
iris.describe()
```

```python
iris.isnull().sum()
```

```python
iris.duplicated().sum()
```

```python
iris.drop_duplicates(inplace=True)
```

```python
sns.pairplot(iris,hue='Species',palette='Set1')
plt.show()
```

```python
X=iris.iloc[:,:4].values
y=iris['target'].values
```

```python
sc=StandardScaler()
X_sc=sc.fit_transform(X)
```

```python
inertia=[]

for k in range(1,10):
  kmeans=KMeans(n_clusters=k,random_state=0,n_init=10)
  kmeans.fit(X_sc)
  inertia.append(kmeans.inertia_)
```

```python
plt.plot(range(1,10),inertia,marker='o')
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
```

```python
sil_score=[]
k_range=range(2,10)

for k in k_range:
  kmeans=KMeans(n_clusters=k,random_state=0,n_init=10)
  labels=kmeans.fit_predict(X_sc)

  score=silhouette_score(X_sc,labels)
  sil_score.append(score)
```

```python
plt.plot(k_range,sil_score,marker='o')
plt.title('Silhouette score')
plt.xlabel('Number of clusters')
plt.ylabel('Silhouette score')
plt.show()
```

```python
best_k=k_range[sil_score.index(max(sil_score))]
print("best k",best_k)
```

```python
kmeans=KMeans(n_clusters=best_k,random_state=0,n_init=10)
labels=kmeans.fit_predict(X_sc)
```

```python
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X_sc)
```

```python
center=pca.transform(kmeans.cluster_centers_)
```

```python
plt.scatter(X_pca[:,0],X_pca[:,1],c=labels,cmap='viridis',s=50)
plt.scatter(center[:,0],center[:,1],c='red',s=200,marker='X')
plt.title('Kmenas clustring with PCA')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.show()
```

```python
final_score=silhouette_score(X_sc,labels)
print("Silhouette score",final_score)
```

New Example: Hierarchical clustering

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import load_iris

from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
iris=load_iris()
```

```python
X=iris.data
y=iris.target
```

```python
df=pd.DataFrame(data=X,columns=iris.feature_names)
df['target']=y
```

```python
df['species']=df['target'].apply(lambda x:iris.target_names[x])
```

```python
df.info()
```

```python
df.head()
```

```python
sc=StandardScaler()
X_sc=sc.fit_transform(X)
```

```python
#apply model
model=linkage(X_sc,method='ward')
dendrogram(model)
```

```python
#apply agglomerative clustering
agg=AgglomerativeClustering(n_clusters=3,metric='euclidean',linkage='ward')
y_labels=agg.fit_predict(X_sc)
```

```python
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X_sc)
```

```python
plt.scatter(X_pca[:,0],X_pca[:,1],c=y_labels,cmap='viridis',s=50)
```

```python
print("Cluster count",pd.Series(y_labels).value_counts())
```
